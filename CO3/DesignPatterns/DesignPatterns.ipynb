# Design Patterns


```python
from abc import ABC, abstractmethod

# Strategy
class Sorter(ABC):
    @abstractmethod
    def sort(self, d): ...
class BubbleSort(Sorter):
    def sort(self, d):
        a=d[:]; [(a.__setitem__(j,a[j+1]) or a.__setitem__(j+1,a[j])) for i in range(len(a)) for j in range(len(a)-i-1) if a[j]>a[j+1]]; return a
class QuickSort(Sorter):
    def sort(self, d): return sorted(d)

# Singleton
class AppConfig:
    _inst=None
    def __new__(cls): cls._inst=cls._inst or super().__new__(cls); return cls._inst

# Observer
class EventBus:
    def __init__(self): self._subs=[]
    def on(self,fn): self._subs.append(fn)
    def emit(self,e): [f(e) for f in self._subs]

data=[5,3,8,1,9]
print("Bubble:", BubbleSort().sort(data))
print("Quick:", QuickSort().sort(data))
c1,c2=AppConfig(),AppConfig(); print("Singleton:",c1 is c2)
bus=EventBus(); bus.on(lambda e: print(f"Event: {e}")); bus.emit("USER_LOGIN")
```
