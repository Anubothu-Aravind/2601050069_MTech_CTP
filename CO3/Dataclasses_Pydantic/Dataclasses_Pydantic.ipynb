# Dataclasses & Pydantic Validation


```python
from dataclasses import dataclass, field
from typing import List

@dataclass(order=True, frozen=True)
class Student:
    roll: int
    name: str
    marks: float = field(compare=False)
    grade: str = field(init=False, compare=False)

    def __post_init__(self):
        object.__setattr__(self,'grade','A' if self.marks>=90 else 'B' if self.marks>=75 else 'C')

students=[Student(3,"Ravi",88.5), Student(1,"Aravind",95.0), Student(2,"Priya",72.3)]
for s in sorted(students): print(f"Roll {s.roll}: {s.name} | Marks: {s.marks} | Grade: {s.grade}")

# Pydantic-style manual validation
@dataclass
class Employee:
    emp_id: int; name: str; salary: float
    def __post_init__(self):
        if self.salary < 0: raise ValueError(f"Salary cannot be negative: {self.salary}")
        if not self.name.strip(): raise ValueError("Name cannot be empty")

e=Employee(101,"Kiran",55000.0); print(f"\nEmployee: {e}")
try: Employee(102,"",−1000)
except ValueError as ex: print(f"Validation Error: {ex}")
```
