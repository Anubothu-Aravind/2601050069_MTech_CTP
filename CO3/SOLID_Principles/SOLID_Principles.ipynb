# SOLID Principles in Python


```python
from abc import ABC, abstractmethod

# S — Single Responsibility
class InvoiceCalculator:
    def total(self, items): return sum(p*q for p,q in items)

class InvoicePrinter:
    def print(self, total): print(f"Invoice Total: Rs.{total:,.2f}")

# O — Open/Closed (extend without modifying)
class Discount(ABC):
    @abstractmethod
    def apply(self, price): ...
class NoDiscount(Discount):
    def apply(self, price): return price
class StudentDiscount(Discount):
    def apply(self, price): return price*0.8

# L — Liskov Substitution
class Bird(ABC):
    @abstractmethod
    def move(self): ...
class Sparrow(Bird):
    def move(self): return "Flying"
class Penguin(Bird):
    def move(self): return "Swimming"

# I — Interface Segregation (separate ABCs)
class Readable(ABC):
    @abstractmethod
    def read(self): ...
class Writable(ABC):
    @abstractmethod
    def write(self, data): ...

# D — Dependency Inversion
class Notifier(ABC):
    @abstractmethod
    def send(self, msg): ...
class EmailNotifier(Notifier):
    def send(self, msg): return f"Email: {msg}"
class SMSNotifier(Notifier):
    def send(self, msg): return f"SMS: {msg}"
class OrderService:
    def __init__(self, notifier: Notifier): self.n=notifier
    def place(self, item): print(self.n.send(f"Order placed: {item}"))

calc=InvoiceCalculator(); printer=InvoicePrinter()
printer.print(calc.total([("Laptop",1,75000),("Mouse",2,500)][0:1]+[(75000,1)]))
for b in [Sparrow(),Penguin()]: print(f"{b.__class__.__name__}: {b.move()}")
OrderService(EmailNotifier()).place("Laptop")
OrderService(SMSNotifier()).place("Mouse")
```
