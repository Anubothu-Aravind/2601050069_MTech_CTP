# Mixins & Protocols (PEP 544)


```python
from typing import Protocol, runtime_checkable

# Mixin — reusable behaviour injected via multiple inheritance
class LogMixin:
    def log(self, msg): print(f"[{self.__class__.__name__}] {msg}")

class JsonMixin:
    def to_json(self):
        import json
        return json.dumps({k:v for k,v in self.__dict__.items()})

class Product(LogMixin, JsonMixin):
    def __init__(self, name, price):
        self.name=name; self.price=price
    def discount(self, pct):
        self.price*=(1-pct/100)
        self.log(f"Price after {pct}% discount: {self.price:.2f}")

# Protocol — structural subtyping (duck typing with type checking)
@runtime_checkable
class Drawable(Protocol):
    def draw(self) -> str: ...

class Circle:
    def draw(self): return "Drawing Circle"
class Triangle:
    def draw(self): return "Drawing Triangle"

p=Product("Laptop",80000)
p.discount(10)
print(p.to_json())

for shape in [Circle(), Triangle()]:
    print(isinstance(shape, Drawable), shape.draw())
```
