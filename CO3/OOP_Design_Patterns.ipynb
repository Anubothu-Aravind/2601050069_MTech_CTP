# CO3 — OOP Design Patterns Demo


```python
from abc import ABC, abstractmethod
from dataclasses import dataclass

# Strategy Pattern
class SortStrategy(ABC):
    @abstractmethod
    def sort(self, data: list) -> list: ...

class BubbleSort(SortStrategy):
    def sort(self, data):
        a = data[:]
        for i in range(len(a)):
            for j in range(len(a)-i-1):
                if a[j] > a[j+1]: a[j], a[j+1] = a[j+1], a[j]
        return a

class PythonSort(SortStrategy):
    def sort(self, data): return sorted(data)

# Singleton Pattern
class Config:
    _instance = None
    def __new__(cls):
        if not cls._instance: cls._instance = super().__new__(cls)
        return cls._instance
    def __init__(self): self.theme = "dark"

# Observer Pattern (minimal)
class EventBus:
    def __init__(self): self._listeners = []
    def subscribe(self, fn): self._listeners.append(fn)
    def publish(self, event): [fn(event) for fn in self._listeners]

# Demo
data = [5, 2, 8, 1, 9]
print("Bubble Sort:", BubbleSort().sort(data))
print("Python Sort:", PythonSort().sort(data))
c1, c2 = Config(), Config()
print("Singleton same?", c1 is c2)
bus = EventBus()
bus.subscribe(lambda e: print(f"Received: {e}"))
bus.publish("ORDER_PLACED")
```
