# Comprehensions vs Loops — Performance


```python
import time

N = 500000

# Loop
t=time.perf_counter()
result=[]
for x in range(N):
    if x%2==0: result.append(x*x)
print(f"Loop:          {time.perf_counter()-t:.4f}s  len={len(result)}")

# Comprehension
t=time.perf_counter()
result=[x*x for x in range(N) if x%2==0]
print(f"Comprehension: {time.perf_counter()-t:.4f}s  len={len(result)}")

# Dict comprehension
squares={x:x*x for x in range(1,11)}
print("Dict comp:", squares)

# Set comprehension
unique={x%5 for x in range(20)}
print("Set comp:", sorted(unique))
```
