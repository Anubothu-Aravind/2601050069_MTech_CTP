# Generators & Iterators


```python
import tracemalloc

def squares_gen(n):
    for i in range(1, n+1):
        yield i*i

# Memory comparison
tracemalloc.start()
lst = [x*x for x in range(1, 100001)]
_, list_mem = tracemalloc.get_traced_memory(); tracemalloc.reset_peak()
gen = (x*x for x in range(1, 100001)); total = sum(gen)
_, gen_mem = tracemalloc.get_traced_memory(); tracemalloc.stop()

print("List memory:", list_mem // 1024, "KB")
print("Generator memory:", gen_mem // 1024, "KB")
print("First 5 squares:", list(squares_gen(5)))
```
