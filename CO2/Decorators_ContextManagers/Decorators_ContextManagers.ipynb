# Decorators & Context Managers


```python
import time, functools
from contextlib import contextmanager

def timer(fn):
    @functools.wraps(fn)
    def wrapper(*args, **kw):
        t = time.perf_counter()
        result = fn(*args, **kw)
        print(f"{fn.__name__} took {time.perf_counter()-t:.4f}s")
        return result
    return wrapper

@timer
def slow_sum(n): return sum(range(n))

@contextmanager
def managed_resource(name):
    print(f"[OPEN] {name}"); yield name; print(f"[CLOSE] {name}")

print("Sum:", slow_sum(1000000))
with managed_resource("DB Connection") as r:
    print(f"Using {r}")
```
