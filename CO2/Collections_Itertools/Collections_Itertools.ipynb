# Collections & Itertools


```python
from collections import Counter, deque, defaultdict
from itertools import islice, chain
import heapq, bisect

freq = Counter("aabbccca")
print("Counter:", freq.most_common(2))
q = deque([1,2,3]); q.appendleft(0)
print("Deque:", list(q))
heap = [5,2,8,1]; heapq.heapify(heap)
print("Min heap top:", heapq.heappop(heap))
arr = [10,20,30,40]; bisect.insort(arr, 25)
print("After insort:", arr)
```
