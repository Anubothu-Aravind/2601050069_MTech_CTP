# CO2 — Pythonic Data Structures Demo


```python
from collections import Counter, deque, defaultdict
from itertools import islice
from functools import lru_cache
import heapq, bisect

# 1. Counter
words = "apple banana apple cherry banana apple"
freq = Counter(words.split())
print("Top 2:", freq.most_common(2))

# 2. deque as queue
q = deque([1, 2, 3])
q.appendleft(0); q.pop()
print("Deque:", list(q))

# 3. heapq – min-heap top-3 salaries
salaries = [45000, 72000, 61000, 89000, 53000]
top3 = heapq.nlargest(3, salaries)
print("Top 3 Salaries:", top3)

# 4. bisect – sorted insertion point
arr = [10, 20, 30, 40, 50]
pos = bisect.bisect_left(arr, 25)
print("Insert 25 at index:", pos)

# 5. Generator – lazy squares
squares = (x*x for x in range(1, 100001))
print("First 5 lazy squares:", list(islice(squares, 5)))

# 6. lru_cache – memoised fibonacci
@lru_cache(maxsize=None)
def fib(n): return n if n < 2 else fib(n-1)+fib(n-2)
print("Fib(30):", fib(30))
```
