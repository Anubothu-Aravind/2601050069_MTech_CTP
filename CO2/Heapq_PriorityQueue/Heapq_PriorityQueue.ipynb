# heapq — Priority Queue Applications


```python
import heapq

# Min-heap task scheduler
tasks=[(3,"Low priority task"),(1,"Critical task"),(2,"Medium task"),(1,"Another critical")]
heapq.heapify(tasks)
print("Task execution order:")
while tasks:
    pri,task=heapq.heappop(tasks)
    print(f"  [{pri}] {task}")

# Top-K elements
scores=[85,92,78,96,88,71,99,83]
top3=heapq.nlargest(3,scores)
bottom3=heapq.nsmallest(3,scores)
print("Top 3 scores:", top3)
print("Bottom 3 scores:", bottom3)

# Merge sorted lists
a,b,c=[1,4,7],[2,5,8],[3,6,9]
merged=list(heapq.merge(a,b,c))
print("Merged sorted:", merged)
```
