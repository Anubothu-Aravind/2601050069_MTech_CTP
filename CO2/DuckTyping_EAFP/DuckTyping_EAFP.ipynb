# Duck Typing & EAFP vs LBYL


```python
# LBYL — Look Before You Leap
def get_len_lbyl(obj):
    if hasattr(obj, '__len__'): return len(obj)
    return "No length"

# EAFP — Easier to Ask Forgiveness than Permission
def get_len_eafp(obj):
    try: return len(obj)
    except TypeError: return "No length"

# Duck typing — any object with .area() works
class Circle:
    def __init__(self,r): self.r=r
    def area(self): return 3.14159*self.r*self.r

class Square:
    def __init__(self,s): self.s=s
    def area(self): return self.s*self.s

shapes=[Circle(5),Square(4),Circle(3)]
for s in shapes:
    print(f"{s.__class__.__name__} area: {s.area():.2f}")

print("LBYL:", get_len_lbyl([1,2,3]), get_len_lbyl(42))
print("EAFP:", get_len_eafp([1,2,3]), get_len_eafp(42))
```
