Given a database of students with their names and marks: 
* Anitha: 95
* Vivek: 83
* Laksmi: 67
* Teja: 97
* Kumar: 85

Sort the list of students in **descending order** of their marks using the **Merge Sort** algorithm, and select the students who are eligible for a scholarship (criteria: marks $\ge 90$).

## 4. Algorithm
The Merge Sort algorithm for descending order operates as follows:
1. **Divide**: If the list has a length of $1$ or less, it is already sorted; return it (base case). Otherwise, divide the list into two halves at the midpoint: `mid = len(arr) // 2`.
2. **Conquer**: Recursively apply Merge Sort to the left half and right half.
3. **Combine (Merge)**: Merge the two sorted halves back into a single sorted list in descending order:
   * Compare elements at the pointers of both halves.
   * If the marks of the student in the left half are greater than or equal to the marks of the student in the right half, insert the left student into the temporary list (this preserves algorithm stability). Otherwise, insert the right student.
   * Copy any remaining students from either half.
4. **Scholarship Filtering**: Traverse the sorted list and select students with `marks >= 90`.

```python
# =========================================
# 5. Core Logic
# =========================================

def merge_sort_students(arr):
    """
    Sorts a list of student tuples (name, marks) in descending order of marks
    using a stable Merge Sort algorithm.
    """
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    left_half = arr[:mid]
    right_half = arr[mid:]

    # Recursive divide and conquer
    left_sorted = merge_sort_students(left_half)
    right_sorted = merge_sort_students(right_half)

    # Merge step (descending order comparison)
    merged = []
    i = j = 0

    while i < len(left_sorted) and j < len(right_sorted):
        # Stable sort: left element comes first if marks are equal
        if left_sorted[i][1] >= right_sorted[j][1]:
            merged.append(left_sorted[i])
            i += 1
        else:
            merged.append(right_sorted[j])
            j += 1

    # Append remaining elements
    merged.extend(left_sorted[i:])
    merged.extend(right_sorted[j:])

    return merged
```

```python
# =========================================
# 6. Data Store & Input Setup
# =========================================

# Given database of students and their marks
students_database = [
    ("Anitha", 95),
    ("Vivek", 83),
    ("Laksmi", 67),
    ("Teja", 97),
    ("Kumar", 85)
]

# Scholarship eligibility criteria (marks >= 90)
scholarship_threshold = 90
```

```python
# =========================================
# 8. Driver Logic (Main)
# =========================================

def run_scholarship_program(students, threshold):
    """
    Orchestrates sorting of students and displays scholarship eligibility results.
    """
    print("\nInitial Student Database:", students)
    
    # Apply Merge Sort
    sorted_students = merge_sort_students(students)
    print("\nStudents Sorted in Descending Order of Marks:")
    for name, mark in sorted_students:
        print(f"  - {name}: {mark}")

    # Filter Scholarship Eligibility
    scholarship_recipients = [student for student in sorted_students if student[1] >= threshold]
    
    print(f"\nScholarship Recipients (Marks >= {threshold}):")
    if scholarship_recipients:
        for name, mark in scholarship_recipients:
            print(f"  * {name} ({mark}) - ELIGIBLE")
    else:
        print("  None of the students are eligible for a scholarship.")
    return sorted_students, scholarship_recipients

def interactive_mode():
    """
    Prompts for custom students input dynamically, handles exceptions for non-interactive runners.
    """
    print("\n=============================================")
    print("        DYNAMIC INPUT / INTERACTIVE MODE")
    print("=============================================")
    
    # Ask for custom scholarship threshold
    try:
        threshold_input = input("Enter scholarship eligibility marks threshold [default: 90]: ").strip()
        threshold = int(threshold_input) if threshold_input else 90
    except Exception:
        print("Non-interactive environment. Using default threshold: 90")
        threshold = 90

    # Ask for custom student count
    try:
        count_input = input("Enter number of custom students to add [default: 0 to skip to preset database]: ").strip()
        count = int(count_input) if count_input else 0
    except Exception:
        count = 0

    students = []
    if count > 0:
        print(f"Please enter {count} student name and marks:")
        for index in range(count):
            try:
                name = input(f"  Student {index+1} Name: ").strip()
                marks = int(input(f"  Student {index+1} Marks: ").strip())
                students.append((name, marks))
            except Exception:
                print("Invalid inputs, using default preset list instead.")
                students = students_database.copy()
                break
    else:
        students = students_database.copy()

    # Run the program with selected database
    run_scholarship_program(students, threshold)
    print("=============================================\n")

# Run the dynamic interactive scholarship application
interactive_mode()
```
