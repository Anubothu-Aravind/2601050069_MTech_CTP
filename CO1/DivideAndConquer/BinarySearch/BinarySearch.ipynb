## 1. Problem Statement
A university library has 1,000,000 (10 lakh) books arranged in increasing order (e.g., 1, 2, ..., 1,000,000). A student wants to find a book with ID `750000`. The system must search for the book, return its position (index/location) if found, or output that the book was "Not Found".

## 2. Divide into Parts/Modules
To solve this problem cleanly, the system is divided into the following functional parts:
1. **Core Search Logic**: A reusable function implementing the binary search algorithm.
2. **Data Store & Input Setup**: Simulates the library database (10 lakh sorted book IDs) and handles query parameter definitions.
3. **Sample Edge Cases (Test Cases)**: Conceptual scenarios to verify the system correctness under extreme inputs.

## 3. Abstractions
Abstraction helps simplify the problem by focusing only on what is necessary and ignoring irrelevant details.

* **Target Book ID**: The specific book ID the student is searching for.
* **Control Pointers**: Indices (`low`, `high`, `mid`) to track and divide the search boundaries.

### Useless Details (Irrelevant Information):

## 4. Algorithm
The Binary Search algorithm operates as follows:
1. Set `low` pointer to the first index ($0$) and `high` pointer to the last index ($N-1$).
2. While `low <= high`:
   * Compute the middle index: `mid = low + (high - low) // 2`.
   * If the book at `books[mid]` matches the target, return `mid` (Found).
   * If `books[mid] < target`, discard the left half by setting `low = mid + 1`.
   * If `books[mid] > target`, discard the right half by setting `high = mid - 1`.
3. If the loop ends and `low > high`, return `-1` (Not Found).

```python
def binary_search_books(books, target_id):
    low = 0
    high = len(books) - 1
    
    while low <= high:
        mid = low + (high - low) // 2
        if books[mid] == target_id:
            return mid
        elif books[mid] < target_id:
            low = mid + 1
        else:
            high = mid - 1
            
    return -1
```

```python
print("=============================================")
print("        University Library")
print("=============================================")

def indian_format(number):
    s = str(number)

    if len(s) <= 3:
        return s

    last_three = s[-3:]
    remaining = s[:-3]

    parts = []

    while len(remaining) > 2:
        parts.insert(0, remaining[-2:])
        remaining = remaining[:-2]

    if remaining:
        parts.insert(0, remaining)

    return ",".join(parts) + "," + last_three

default_N = 1000000

user_input = input(
    f"\nEnter number of books [default: {indian_format(default_N)}]: "
)

N = int(user_input) if user_input else default_N

books = list(range(1, N + 1))
continue_search = True

while continue_search:

    default_target = 750000
    user_input = input(
        f"\nEnter Book ID to search [default: {indian_format(default_target)}]: "
    )
    target = int(user_input) if user_input else default_target
    index = binary_search_books(books, target)
    print("\n=============================================")
    print("        University Library - Search Results")
    print("=============================================")

    if index != -1:
        print(f"SUCCESS: Book {indian_format(target)} found!")
        print(f"Index: {indian_format(index)}")
        print(f"Position: {indian_format(index + 1)}")
    else:
        print(
            f"NOT FOUND: Book {indian_format(target)} "
            f"is not in the library."
        )

    print("=============================================")
    while True:

        choice = input(
            "\nDo you want to search another book? (yes/no): "
        ).strip().lower()

        if choice in ["yes", "y"]:
            break

        elif choice in ["no", "n"]:
            continue_search = False
            print("\nThank you for using the University Library!")
            break

        else:
            print("Invalid input. Please enter yes/y or no/n.")
```
