## 1. Problem Statement
A university library has 1,000,000 (10 lakh) books arranged in increasing order (e.g., 1, 2, ..., 1,000,000). A student wants to find a book with ID `750000`. The system must search for the book, return its position (index/location) if found, or output that the book was "Not Found".

## 4. Algorithm
The Binary Search algorithm operates as follows:
1. Set `low` pointer to the first index ($0$) and `high` pointer to the last index ($N-1$).
2. While `low <= high`:
   * Compute the middle index: `mid = low + (high - low) // 2`.
   * If the book at `books[mid]` matches the target, return `mid` (Found).
   * If `books[mid] < target`, discard the left half by setting `low = mid + 1`.
   * If `books[mid] > target`, discard the right half by setting `high = mid - 1`.
3. If the loop ends and `low > high`, return `-1` (Not Found).

```python
def binary_search_books(books, target_id):
    low = 0
    high = len(books) - 1
    
    while low <= high:
        mid = low + (high - low) // 2
        if books[mid] == target_id:
            return mid
        elif books[mid] < target_id:
            low = mid + 1
        else:
            high = mid - 1
            
    return -1
```

```python
def interactive_mode():
    print("\n=============================================")
    print("        University Library")
    print("=============================================")
    
    try:
        n_input = input("Enter the total number of books in the library [default: 1000000]: ").strip()
        N = int(n_input) if n_input else 1000000
    except Exception:
        print("Non-interactive environment detected. Using default library size: 1000000")
        N = 1000000
        
    if N < 0:
        print("Error: Number of books cannot be negative.")
        return
        
    print(f"Generating sorted library of {N} books (IDs 1 to {N})...")
    books = list(range(1, N + 1))
    
    try:
        target_input = input(f"Enter the book ID to search for (1 to {N}) [default: 750000]: ").strip()
        target = int(target_input) if target_input else 750000
    except Exception:
        print("Non-interactive environment detected. Using default target ID: 750000")
        target = 750000
        
    print(f"Searching for book {target} using Binary Search...")
    index = binary_search_books(books, target)
    
    if index != -1:
        print(f"\nSUCCESS: Book {target} found at index {index} (Position {index + 1} in the sorted list).")
        return index
    else:
        print(f"\nNOT FOUND: Book {target} is not in the library database.")
        return -1
    print("=============================================\n")

# Execute the driver logic
interactive_mode()
```
