# Effective Prompting for Code Generation


```python
# AI Pair Programming — Prompt Engineering Best Practices

techniques = {
    "Specification-First": {
        "bad":  "Write a sort function",
        "good": "Write merge_sort(arr: list[int]) -> list[int] that sorts ascending. Add docstring, type hints, return sorted copy without mutating input. Time: O(n log n), Space: O(n)."
    },
    "Context Injection": {
        "bad":  "Fix this bug",
        "good": "In binary_search() below, the loop exits without returning -1 when target not found. Fix only the exit condition. [paste function]"
    },
    "Test-First Prompting": {
        "bad":  "Write a function that validates email",
        "good": "Write validate_email(email: str) -> bool that passes: validate_email('a@b.com')==True, validate_email('bad')==False, validate_email('')==False."
    },
    "Refactoring Prompt": {
        "bad":  "Make this code better",
        "good": "Refactor to use dataclasses instead of dicts, add type hints, extract magic numbers as named constants. Keep same behaviour."
    },
}

for name, tips in techniques.items():
    print(f"\n{'='*50}")
    print(f"TECHNIQUE: {name}")
    print(f"  BAD  PROMPT: {tips['bad']}")
    print(f"  GOOD PROMPT: {tips['good'][:80]}...")
```
