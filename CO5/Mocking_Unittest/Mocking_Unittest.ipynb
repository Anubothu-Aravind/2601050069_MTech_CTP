# Mocking with unittest.mock


```python
from unittest.mock import MagicMock, patch, call

# Mock external API
class PaymentGateway:
    def charge(self, amount): pass  # real call hits network

class OrderService:
    def __init__(self, gateway): self.gw=gateway
    def place_order(self, item, price):
        result=self.gw.charge(price)
        return f"Order for {item}: {'SUCCESS' if result else 'FAILED'}"

mock_gw=MagicMock()
mock_gw.charge.return_value=True
svc=OrderService(mock_gw)

print(svc.place_order("Laptop",75000))
print(svc.place_order("Mouse",500))

mock_gw.charge.assert_called_with(500)
print("Charge call args:", mock_gw.charge.call_args_list)

# patch context manager
with patch("builtins.print") as mock_print:
    print("this is captured")
    mock_print.assert_called_with("this is captured")
    print("Mock captured print call:", mock_print.call_count, "times")
```
