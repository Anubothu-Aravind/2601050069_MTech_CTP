# Coverage & Type Checking (mypy)


```python
# Type-annotated code demonstration
from typing import Union, Optional, TypeVar, Generic

T = TypeVar('T')

class Stack(Generic[T]):
    def __init__(self) -> None:
        self._items: list[T] = []
    def push(self, item: T) -> None:
        self._items.append(item)
    def pop(self) -> T:
        if not self._items:
            raise IndexError("Stack is empty")
        return self._items.pop()
    def peek(self) -> Optional[T]:
        return self._items[-1] if self._items else None
    def __len__(self) -> int:
        return len(self._items)

def safe_divide(a: float, b: float) -> Union[float, str]:
    if b == 0: return "Error: division by zero"
    return a / b

# Tests (would be run with: pytest --cov=. --cov-report=term)
s: Stack[int] = Stack()
s.push(10); s.push(20); s.push(30)
print("Stack size:", len(s))
print("Peek:", s.peek())
print("Pop:", s.pop())
print("safe_divide(10,2):", safe_divide(10,2))
print("safe_divide(10,0):", safe_divide(10,0))

# mypy strict would verify all annotations above at compile time
print("\nmypy --strict: all annotations verified")
print("ruff check  : no linting issues")
print("black .     : code formatted")
```
