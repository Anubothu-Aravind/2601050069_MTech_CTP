```python
# Unit tests inline (normally run via: pytest)
import traceback

def add(a: int, b: int) -> int:
    return a + b

def divide(a: float, b: float) -> float:
    if b == 0:
        raise ValueError("Division by zero")
    return a / b

# Simple test runner
passed = failed = 0
def test(name, fn):
    global passed, failed
    try: fn(); print(f"  PASS  {name}"); passed += 1
    except AssertionError as e: print(f"  FAIL  {name}: {e}"); failed += 1

print("=== Running Unit Tests ===")
test("add positive",   lambda: assert_eq(add(2, 3), 5))
test("add negative",   lambda: assert_eq(add(-1, 1), 0))
test("divide normal",  lambda: assert_eq(divide(10, 2), 5.0))
test("divide by zero", lambda: assert_raises(ValueError, divide, 1, 0))

def assert_eq(a, b): assert a == b, f"{a} != {b}"
def assert_raises(exc, fn, *args):
    try: fn(*args); assert False, "Expected exception"
    except exc: pass

print("=== Running Unit Tests ===")
test("add positive",   lambda: assert_eq(add(2, 3), 5))
test("add negative",   lambda: assert_eq(add(-1, 1), 0))
test("divide normal",  lambda: assert_eq(divide(10, 2), 5.0))
test("divide by zero", lambda: assert_raises(ValueError, divide, 1, 0))

print(f"\nResults: {passed} passed, {failed} failed")
```
