# Testing with Pytest & Hypothesis


```python
def binary_search(arr, target):
    lo,hi = 0,len(arr)-1
    while lo<=hi:
        mid=(lo+hi)//2
        if arr[mid]==target: return mid
        elif arr[mid]<target: lo=mid+1
        else: hi=mid-1
    return -1

# Inline test suite
tests = [
    ("found middle",   [1,3,5,7,9], 5, 2),
    ("found first",    [1,3,5,7,9], 1, 0),
    ("found last",     [1,3,5,7,9], 9, 4),
    ("not found",      [1,3,5,7,9], 4, -1),
    ("empty array",    [],          1, -1),
]
p=f=0
for name,arr,tgt,exp in tests:
    got=binary_search(arr,tgt)
    if got==exp: print(f"  PASS {name}"); p+=1
    else: print(f"  FAIL {name}: expected {exp} got {got}"); f+=1

print(f"\n{p} passed, {f} failed")
```
