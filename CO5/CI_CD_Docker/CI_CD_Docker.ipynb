# CI/CD & Docker Basics


```python
# GitHub Actions workflow (YAML preview)
workflow = """
name: CI
on: [push]
jobs:
  test:
    runs-on: ubuntu-latest
    steps:
      - uses: actions/checkout@v4
      - uses: actions/setup-python@v5
        with: { python-version: '3.12' }
      - run: pip install -e .[dev]
      - run: ruff check .
      - run: mypy --strict src/
      - run: pytest --cov=src --cov-report=term-missing
"""

dockerfile = """
FROM python:3.12-slim
WORKDIR /app
COPY pyproject.toml .
RUN pip install -e .
COPY src/ src/
CMD ["python", "-m", "mypackage"]
"""

print("=== .github/workflows/ci.yml ===")
print(workflow)
print("=== Dockerfile ===")
print(dockerfile)
```
