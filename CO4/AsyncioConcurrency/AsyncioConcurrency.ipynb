# Asyncio Concurrency


```python
import asyncio, time

async def fetch(sem, id_):
    async with sem:
        await asyncio.sleep(0.05)
        return f"data-{id_}"

async def main():
    sem = asyncio.Semaphore(5)
    t = time.perf_counter()
    results = await asyncio.gather(*[fetch(sem,i) for i in range(1,21)])
    print(f"20 tasks in {time.perf_counter()-t:.3f}s (sequential would take 1.0s)")
    print("Sample:", results[:3])

asyncio.run(main())
```
