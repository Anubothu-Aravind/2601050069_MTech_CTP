# Threading vs Multiprocessing — GIL Analysis


```python
import threading, multiprocessing, time

def cpu_task(n):
    return sum(i*i for i in range(n))

N=500000; WORKERS=4

# Sequential
t=time.perf_counter()
[cpu_task(N) for _ in range(WORKERS)]
seq=time.perf_counter()-t; print(f"Sequential:      {seq:.3f}s")

# Threading (GIL blocks CPU parallelism)
t=time.perf_counter()
threads=[threading.Thread(target=cpu_task,args=(N,)) for _ in range(WORKERS)]
[t_.start() for t_ in threads]; [t_.join() for t_ in threads]
thr=time.perf_counter()-t; print(f"Threading:       {thr:.3f}s (GIL — no speedup for CPU)")

# Multiprocessing (true parallelism)
t=time.perf_counter()
with multiprocessing.Pool(WORKERS) as pool:
    pool.map(cpu_task,[N]*WORKERS)
mp=time.perf_counter()-t; print(f"Multiprocessing: {mp:.3f}s")
print(f"Speedup:         {seq/mp:.2f}x")
```
