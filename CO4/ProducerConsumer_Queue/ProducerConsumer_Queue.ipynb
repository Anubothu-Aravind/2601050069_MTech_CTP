# Producer-Consumer Pattern with asyncio Queue


```python
import asyncio, random, time

async def producer(q: asyncio.Queue, n: int, name: str):
    for i in range(n):
        item=f"{name}-item{i+1}"
        await asyncio.sleep(random.uniform(0.01,0.05))
        await q.put(item)
        print(f"  Produced: {item}")
    await q.put(None)  # sentinel

async def consumer(q: asyncio.Queue, name: str):
    while True:
        item=await q.get()
        if item is None: q.task_done(); break
        await asyncio.sleep(0.02)
        print(f"    Consumed [{name}]: {item}")
        q.task_done()

async def main():
    q=asyncio.Queue(maxsize=5)
    t=time.perf_counter()
    await asyncio.gather(
        producer(q,4,"P1"),
        producer(q,3,"P2"),
        consumer(q,"C1"),
        consumer(q,"C2"),
    )
    print(f"Done in {time.perf_counter()-t:.3f}s")

asyncio.run(main())
```
