# Functional Programming


```python
from functools import reduce, partial

nums = list(range(1,11))
evens  = list(filter(lambda x: x%2==0, nums))
squares= list(map(lambda x: x**2, evens))
total  = reduce(lambda a,b: a+b, squares)
print("Even squares sum:", total)

power = lambda base,exp: base**exp
square=partial(power,exp=2); cube=partial(power,exp=3)
print("square(7):", square(7), "cube(4):", cube(4))

# Currying
def curry_add(a): return lambda b: a+b
add5 = curry_add(5)
print("add5(3):", add5(3), "add5(10):", add5(10))
```
