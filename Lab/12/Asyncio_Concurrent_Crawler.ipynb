# Lab 12: Asynchronous I/O and Concurrency using asyncio

## Objective
Implement high-throughput concurrent I/O using Python asyncio, coroutines, async/await, Semaphore rate limiting, and an asynchronous producer-consumer queue pattern.


```python
import asyncio
import time
from typing import List, Dict

async def fetch_resource(sem: asyncio.Semaphore, resource_id: int) -> Dict[str, any]:
    async with sem:
        # Simulate non-blocking network I/O
        await asyncio.sleep(0.05)
        return {
            "resource_id": resource_id,
            "status": 200,
            "data": f"Payload for record #{resource_id}"
        }

async def run_async_batch(total_items: int = 20, max_concurrency: int = 5):
    sem = asyncio.Semaphore(max_concurrency)
    start_time = time.perf_counter()
    tasks = [fetch_resource(sem, i) for i in range(1, total_items + 1)]
    results = await asyncio.gather(*tasks)
    elapsed = time.perf_counter() - start_time
    return results, elapsed

# Run event loop demonstration
results, duration = asyncio.run(run_async_batch(20, 5))
print(f"Processed {len(results)} requests concurrently.")
print(f"Max Concurrent Workers: 5")
print(f"Total Async Duration: {duration:.3f} seconds")
print(f"Equivalent Sequential Duration would be: {20 * 0.05:.3f} seconds")
print(f"Speedup Factor: {(20 * 0.05) / duration:.2f}x")
print("Sample Result:", results[0])
```
