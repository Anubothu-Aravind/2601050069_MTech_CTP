# Lab 7: Job Sequencing with Deadlines using Greedy Algorithm

## Objective
Implement Job Sequencing with Deadlines using the Greedy method to maximize total profit while scheduling each job within its deadline.


```python
from dataclasses import dataclass
from typing import List, Optional

@dataclass
class Job:
    id: str
    deadline: int
    profit: int

def job_sequencing(jobs: List[Job]) -> dict:
    # 1. Sort all jobs in descending order of profit (Greedy criterion)
    sorted_jobs = sorted(jobs, key=lambda j: j.profit, reverse=True)
    
    # 2. Find the maximum deadline to determine available time slots
    max_deadline = max(job.deadline for job in jobs) if jobs else 0
    slots: List[Optional[str]] = [None] * (max_deadline + 1)
    
    total_profit = 0
    scheduled_jobs = []
    
    # 3. Schedule each job in the latest available slot <= deadline
    for job in sorted_jobs:
        for t in range(min(max_deadline, job.deadline), 0, -1):
            if slots[t] is None:
                slots[t] = job.id
                total_profit += job.profit
                scheduled_jobs.append((job.id, t, job.profit))
                break
                
    return {
        "scheduled_sequence": [slots[t] for t in range(1, max_deadline + 1) if slots[t] is not None],
        "slot_assignments": {t: slots[t] for t in range(1, max_deadline + 1) if slots[t] is not None},
        "total_profit": total_profit,
        "scheduled_jobs": scheduled_jobs
    }

# Test sample jobs
sample_jobs = [
    Job("J1", 2, 100),
    Job("J2", 1, 19),
    Job("J3", 2, 27),
    Job("J4", 1, 25),
    Job("J5", 3, 15),
]

result = job_sequencing(sample_jobs)
print("=== Job Sequencing Results ===")
print("Jobs:", [(j.id, f"Deadline: {j.deadline}", f"Profit: {j.profit}") for j in sample_jobs])
print("Optimal Execution Order:", result["scheduled_sequence"])
print("Slot Allocation:", result["slot_assignments"])
print(f"Maximum Total Profit: Rs. {result['total_profit']}")
```
