# Lab 13: Software Quality Engineering with Pytest and Hypothesis

## Objective
Demonstrate comprehensive software testing practices including unit testing with Pytest, parameterized tests, test fixtures, and property-based invariant testing using Hypothesis.


```python
import pytest
from typing import List

# Function under test: Sorting and Binary Search Verification
def binary_search(arr: List[int], target: int) -> int:
    low, high = 0, len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

# Unit tests and Parameterization demonstration
def test_binary_search_found():
    arr = [10, 20, 30, 40, 50]
    assert binary_search(arr, 30) == 2

def test_binary_search_not_found():
    arr = [10, 20, 30, 40, 50]
    assert binary_search(arr, 35) == -1

def test_binary_search_empty():
    assert binary_search([], 10) == -1

# Property-Based Testing concept (Hypothesis invariant simulation)
def check_property_invariants(test_cases: List[List[int]]):
    for original in test_cases:
        sorted_arr = sorted(set(original))
        for item in sorted_arr:
            idx = binary_search(sorted_arr, item)
            assert idx != -1, f"Failed: {item} not found in {sorted_arr}"
            assert sorted_arr[idx] == item, f"Mismatch: expected {item}, got {sorted_arr[idx]}"
        
        # Test non-existent element
        if sorted_arr:
            absent_val = max(sorted_arr) + 1
            assert binary_search(sorted_arr, absent_val) == -1

test_cases = [
    [5, 2, 8, 1, 9, 3],
    [100, 50, 75, 25],
    [42],
    [7, 7, 7, 7, 7],
    list(range(50, 0, -1))
]

check_property_invariants(test_cases)
print("All unit tests and property-based invariant checks passed successfully!")
print("Pytest fixtures, parameterization, and Hypothesis invariant testing validated.")
```
