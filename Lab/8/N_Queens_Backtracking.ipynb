# Lab 8: N-Queens Problem using Backtracking

## Objective
Solve the classic N-Queens puzzle using depth-first backtracking with state-space pruning, ensuring no two queens share a row, column, or diagonal.


```python
from typing import List

def solve_n_queens(n: int) -> List[List[str]]:
    solutions = []
    board = [["." for _ in range(n)] for _ in range(n)]
    
    cols = set()
    diag_pos = set()  # (r + c)
    diag_neg = set()  # (r - c)
    
    def backtrack(r: int):
        if r == n:
            solutions.append(["".join(row) for row in board])
            return
            
        for c in range(n):
            if c in cols or (r + c) in diag_pos or (r - c) in diag_neg:
                continue
                
            cols.add(c)
            diag_pos.add(r + c)
            diag_neg.add(r - c)
            board[r][c] = "Q"
            
            backtrack(r + 1)
            
            # Undo move
            cols.remove(c)
            diag_pos.remove(r + c)
            diag_neg.remove(r - c)
            board[r][c] = "."
            
    backtrack(0)
    return solutions

def print_board(solution: List[str]):
    for row in solution:
        print(" ".join(row))
    print()

N = 4
solutions = solve_n_queens(N)
print(f"Total distinct solutions for {N}-Queens: {len(solutions)}\n")
for idx, sol in enumerate(solutions, 1):
    print(f"--- Solution {idx} ---")
    print_board(sol)
```
